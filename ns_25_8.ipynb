Implement Local and Global Network Properties. Also, measure average path length,
diameter, and network density.   

```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import deque
```

```python
edges = [('A','B'), ('A','C'), ('B','D'),('C','D'), ('D','E'), ('E','F')]

G = nx.Graph()
G.add_edges_from(edges)

nx.draw(G, with_labels=True)
plt.show()
```

```python
def bfs(graph, start):
    distance = {node: -1 for node in graph.nodes}
    distance[start] = 0

    queue = deque([start])

    while queue:
        current = queue.popleft()

        for neighbour in graph[current]:
            if distance[neighbour] == -1:
                distance[neighbour] = distance[current] + 1
                queue.append(neighbour)

    return distance
```

```python
print("Local Network Property")
for node in G.nodes:
    print("Degree of", node, "=", len(G[node]))
```

```python
distances = []
diameter = 0

nodes = list(G.nodes)

for i in range(len(nodes)):
    d = bfs(G, nodes[i])

    for j in range(i + 1, len(nodes)):
        distances.append(d[nodes[j]])

        if d[nodes[j]] > diameter:
            diameter = d[nodes[j]]
```

```python
average_path = sum(distances) / len(distances)
```

```python
n = len(nodes)
m = len(edges)

density = (2 * m) / (n * (n - 1))
```

```python
print("\nGlobal Network Properties")
print("Average Path Length =", round(average_path, 2))
print("Diameter =", diameter)
print("Network Density =", round(density, 2))
print("Density (%) =", round(density * 100, 2))
```
