```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G=nx.Graph()
G=nx.complete_graph(5)
```

```python
print("nodes ",G.nodes())
print("edges ",G.edges())
print("adjacent nodes ",G.adj[1])
print("degree ",G.degree[1])
print("edges of 2 and 3 ",G.edges([2,3]))
print("degree of 2 and 3 ",G.degree([2,3]))
nx.draw(G, with_labels=True)
plt.show()
```

### Decorative Graph Visualization

```python
node_colors = ['red', 'blue', 'green', 'purple', 'orange']
edge_colors = ['lightgray'] * len(G.edges())
node_labels = {0: 'Alpha', 1: 'Beta', 2: 'Gamma', 3: 'Delta', 4: 'Epsilon'}

plt.figure(figsize=(8, 6))
nx.draw(G, with_labels=True, labels=node_labels, node_color=node_colors, node_size=1000, edge_color=edge_colors, width=2, font_size=10, font_color='black' )
plt.title('Complete Graph with Decorative Elements')
plt.show()
```
