Practical 1: Create and Visualise a Simple Network
Using Python and NetworkX, create an undirected social network of at least 8 students. Add
appropriate edges to represent friendships or interactions. Visualize the network using Matplotlib
and calculate the number of nodes, number of edges, degree of each node, degree centrality, and
network density.

```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
G = nx.Graph()

students = ["A", "B", "C", "D", "E", "F", "G", "H"]
G.add_nodes_from(students)

edges = [
    ("A", "B"), ("A", "C"), ("A", "D"),
    ("B", "C"), ("B", "E"),
    ("C", "F"), ("D", "E"),
    ("E", "F"), ("F", "G"),
    ("G", "H"), ("H", "A")
]
G.add_edges_from(edges)
```

```python
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color="skyblue",
        node_size=1000, edge_color="gray")
plt.title("Student Social Network")
plt.show()
```

```python
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())

print("\nDegree of each node:")
for node in G:
    print(node, ":", G.degree(node))

print("\nDegree Centrality:")
print(nx.degree_centrality(G))

print("\nNetwork Density:", nx.density(G))
```

Practical 2: Visualise Networks Using User Attributes
Using NetworkX, create a network of at least 8 students and assign each student attributes such
as age, department, and area of interest. Visualize the network such that node color represents
department/interest and node size represents age. Add a suitable legend and explain how the
attributes are represented in the visualization.

```python
G = nx.Graph()

# Student attributes
students = {
    "A": {"age": 20, "department": "CSE", "interest": "AI"},
    "B": {"age": 21, "department": "CSE", "interest": "Web"},
    "C": {"age": 19, "department": "ECE", "interest": "Robotics"},
    "D": {"age": 22, "department": "ECE", "interest": "AI"},
    "E": {"age": 20, "department": "IT", "interest": "Web"},
    "F": {"age": 21, "department": "IT", "interest": "AI"},
    "G": {"age": 19, "department": "CSE", "interest": "Robotics"},
    "H": {"age": 22, "department": "IT", "interest": "Web"}
}
```

```python
for node, attr in students.items():
    G.add_node(node, **attr)

G.add_edges_from([
    ("A", "B"), ("A", "C"), ("A", "D"),
    ("B", "E"), ("C", "F"), ("D", "E"),
    ("E", "F"), ("F", "G"), ("G", "H"),
    ("H", "A")
])

colors = {"CSE": "red","ECE": "blue","IT": "green"}

node_colors = []
node_sizes = []
```

```python
for node in G:
    node_colors.append(colors[G.nodes[node]["department"]])
    node_sizes.append(G.nodes[node]["age"] * 50)
```

```python
pos = nx.spring_layout(G, seed=42)

fig, ax = plt.subplots()

nx.draw(
    G, pos,
    ax=ax,
    with_labels=True,
    node_color=node_colors,
    node_size=node_sizes,
    edge_color="gray"
)

legend = [
    Patch(facecolor="red", label="CSE"),
    Patch(facecolor="blue", label="ECE"),
    Patch(facecolor="green", label="IT")
]
ax.legend(handles=legend, title="Department")
ax.set_title("Student Network with Attributes")
plt.show()
```
